# **FILE HANDLING IN PYTHON**

- File Handling is the process of creating, reading, writing, updating, and deleting files using Python.
- File Handling allows us to store data permanently.

## File Handling

```
Open File
     ↓
Read / Write / Append
     ↓
Close File
```

### Opening a File

Python provides the open() function for opening files.



```
file = open("filename", "mode")
```



````python
f=open("data.txt","r")     # OPENING A FILE IN READ MODE
                           # IF THE FILE DOESN'T EXIST IT WILL SHOW AN ERROR
                           # BECAUSE WE CAN ONLY READ AN EXISTING FILE
f
````

````python
f.close()                   # ALWAYS CLOSE THE FILE AFTER USE
````

* WE CAN OPEN A FILE USING **_with_** STATEMENT

* IF WE ARE USING _with open()_ WE DONT NEED TO EXPLICITLY CLOSING THE FILE

* IT WILL BE AUTOMATICALLY CLOSED AFTER ALL FILE OPERATION

````python
with open("data.txt","r") as f:    #OPENING A FILE USING with open()
    pass
````

### Why do we close a file?
It

- saves changes
- frees memory
- releases the file

# File Modes



| Mode | File Exists           | File Doesn't Exist | Old Data       |
| ---- | --------------------- | ------------------ | -------------- |
| `r`  | Opens file            | Error              | Safe           |
| `w`  | Opens and clears file | Creates file       | Deleted        |
| `a`  | Opens file            | Creates file       | Preserved      |
| `x`  | Error                 | Creates new file   | Not applicable |


### (1) Reading Files

To read a file we need to open it in read mode.

##### `read()` — This method reads entire file




````python
with open("data.txt",'r') as f:
    content=f.read()
    print(content)
````

````python
with open("script.py", "r") as f:
    code = f.read()
    print(code)
````

````python
with open("script (1).py", "r") as f:
    code = f.read()
    print(code)
````

##### `readline()` — Read one line

readline() reads a single line in the file.

````python
with open("data.txt",'r') as f:
    line=f.readline()
    print(line)
````

* If we are using multiple times it will read each line one by one and will reach the end of the file.

````python
with open("data.txt",'r') as f:
    line1=f.readline()
    line2=f.readline()
    line3=f.readline()
    line4=f.readline()
    print(f"LINE1:{line1}LINE2:{line2}LINE3: {line3}")
    print(f"EMPTY LINE: {line4}")
````

##### `readlines()` — Read all lines into a list

````python
with open("data.txt", "r") as f:
    lines = f.readlines()
    print(lines)
````

##### Reading using Loop

````python
for i in lines:          # ITERATING TO THE LINE LIST
    print(i)
````

````python
file = open("student.txt", "r")

for line in file:
    print(line)
file.close()
````

### (2) Writing Files

* If we are opening a file in write(w) mode, if the file already existing it will overwrite the file.
That means if the file is already containing any text , it will erase all those and write the new contents.

* If the file doesn't exist, it will create a new file with the specified name and will write into it.

````python
with open("data1.txt",'w') as f:     # OPENING FILE IN WRITE MODE
    pass
````

##### Using write()

* If file exists, it overwrites content.

````python
with open("notes.txt", "w") as f:
    f.write("Python File Handling is powerful!\n")
````

````python
with open("hello.py", "w") as file:
    file.write('print("Hello, World!")')
````

````python
with open("calculator.py", "w") as file:
    file.write("num1 = 10\n")
    file.write("num2 = 20\n")
    file.write("sum = num1 + num2\n")
    file.write('print("Sum =", sum)\n')
````

````python
code = """
name = input("Enter your name: ")
print("Welcome", name)
"""

with open("program.py", "w") as file:
    file.write(code)
````

##### Using `writelines()`

Writes multiple lines at once (expects an iterable like a list).

````python
lines = ["First line\n", "Second line\n", "Third line\n"]
with open("notes.txt", "w") as f:
    f.writelines(lines)
````

````python
lines=("First line", "Second line", "Third line")
with open("notes.txt", "w") as f:
    f.writelines(lines)
````

````python
lines={"First line\n", "Second line\n", "Third line\n"}
with open("notes.txt", "w") as f:
    f.writelines(lines)
````

### (3) Appending to Files

* Append means
Add new data at the end.
Existing data is not deleted.



* If the file is not existing, then it will create a new file and will append the contents into it.

````python
with open("data1.txt",'a') as f:
    pass
````

* To add new data without deleting old content we are using 'a' mode.

````python
with open("notes.txt", "a") as f:
    f.write("Adding more lines\n")
````

### File Cursor Operations

* Python keeps a cursor (pointer) to track where in the file operations happen.

#####  tell() — get current cursor position

````python
with open("data.txt", "r")as f:
     print(f.tell())
````

#####  seek(offset, from_what) — move cursor

from_what can be:

0 – from start (default)

1 – from current position

````python
with open("data.txt", "r")as f:
     print(f.seek(5,0))
     print(f.tell())
     data=f.readline(4)
     print(data)
````

````python
with open("data.txt", "r")as f:
     print(f.seek(5))
     print(f.tell())
     data=f.read(15)
     print(data)
````

### Working with Binary Files

````python
with open("image.jpg", "rb") as f:     # reads the file in bytes
    data = f.read()

with open("copy.jpg", "wb") as f:      # write bytes into a new file
    f.write(data)
````

7. Write a Python program to count lines, words, and characters in a text file.

````python
with open("demo.txt", "r") as f:
    lines = f.readlines()

line_count = len(lines)
word_count = sum(len(line.split()) for line in lines)
char_count = sum(len(line) for line in lines)

print("Lines:", line_count)
print("Words:", word_count)
print("Characters:", char_count)
````

9. Demonstrate the use of seek() and tell() in file handling.

````python
with open("demo.txt", "r") as f:
    print("Initial position:", f.tell())
    data = f.read(5)
    print("After reading 5 chars:", f.tell())
    f.seek(0)
    print("After seek(0):", f.tell())
````
